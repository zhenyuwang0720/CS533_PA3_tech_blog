# Reproducible Python Data Analysis
A short notebook showing Pandas rolling statistics on synthetic observations.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({'step': range(1, 8), 'value': [3, 5, 4, 7, 8, 6, 9]})
data['rolling_mean_3'] = data['value'].rolling(3).mean()
display(data)
```

```python
ax = data.plot(x='step', y=['value', 'rolling_mean_3'], marker='o', figsize=(8, 4))
ax.set(title='Synthetic data: values and rolling mean', ylabel='Value')
plt.tight_layout()
plt.show()
```

## Interpretation
The three-observation rolling mean smooths the fluctuations in this **synthetic** sequence. Replace the demonstration data and document the experimental assumptions for real research.
